```python
# Operasi array vs skalar (angka tunggal)
nilai = np.array([75, 85, 90, 65, 80])

print(nilai + 5)
# [80 90 95 70 85] → semua ditambah 5

print(nilai * 2)
# [150 170 180 130 160]

print(nilai / 100)
# [0.75 0.85 0.9  0.65 0.8]

print(nilai ** 2)
# [5625 7225 8100 4225 6400]


# Operasi antar dua array (elemen per elemen)
uts = np.array([70, 80, 85, 60, 75])
uas = np.array([80, 90, 95, 70, 85])

nilai_akhir = (uts * 0.4) + (uas * 0.6)

print(nilai_akhir)
# [76. 86. 91. 66. 81.]
```

```python
# TUGAS PERTEMUAN 4
# ARRAY DATA TINGGI BADAN 10 SISWA

import numpy as np


# DATA SISWA

# Array 1D tinggi badan 10 siswa (cm)
data_tinggi = np.array([
    165, 170, 158, 172, 162,
    175, 168, 180, 155, 170
])

# Array 1D berat badan 10 siswa (kg)
data_berat = np.array([
    58, 63, 50, 68, 55,
    72, 60, 75, 48, 65
])

# Array 2D: 10 siswa x 2 kolom
# Kolom 0 = tinggi
# Kolom 1 = berat
data_siswa = np.array([
    [165, 58],
    [170, 63],
    [158, 50],
    [172, 68],
    [162, 55],
    [175, 72],
    [168, 60],
    [180, 75],
    [155, 48],
    [170, 65]
])

# 1. ATRIBUT ARRAY 2D

print("ATRIBUT ARRAY DATA SISWA")
print("Shape :", data_siswa.shape)
print("Ndim  :", data_siswa.ndim)
print("Dtype :", data_siswa.dtype)
print("Size  :", data_siswa.size)


# 2. INDEXING

print()
print("INDEXING")

# Tinggi badan siswa ke-1 (index 0)
print("Tinggi siswa ke-1 :", data_tinggi[0], "cm")

# Berat badan siswa ke-7 (index 6)
print("Berat siswa ke-7  :", data_berat[6], "kg")

# Data lengkap siswa ke-5 (index 4)
print("Data siswa ke-5   :", data_siswa[4])


# 3. SLICING

print()
print("SLICING")

# Tinggi badan 5 siswa pertama
print("Tinggi 5 siswa pertama :", data_tinggi[:5])

# Berat badan 3 siswa terakhir
print("Berat 3 siswa terakhir :", data_berat[-3:])

# Data siswa ke-3 sampai ke-7
print("Data siswa ke-3 sampai ke-7 :")
print(data_siswa[2:7])

# 4. OPERASI ARITMATIKA DAN KATEGORI IMT

# Mengubah tinggi dari cm menjadi meter
tinggi_m = data_tinggi / 100

# Menghitung IMT semua siswa
imt = data_berat / (tinggi_m ** 2)

# Menentukan kategori IMT menggunakan if-elif-else
kategori = []

for nilai in imt:
    if nilai < 18.5:
        kategori.append("Kurus")
    elif nilai < 25:
        kategori.append("Normal")
    elif nilai < 30:
        kategori.append("Gemuk")
    else:
        kategori.append("Obesitas")

# 5. TABEL REKAP

print()
print("REKAP DATA SISWA + IMT")
print("No  Tinggi  Berat   IMT    Kategori")
print("--------------------------------------")

for i in range(10):
    print(
        f"{i+1:<3} "
        f"{data_tinggi[i]:<7} "
        f"{data_berat[i]:<7} "
        f"{imt[i]:<6.2f} "
        f"{kategori[i]}"
    )
```
